# Preprocessing

```python
x = df.drop('music_genre', axis= 1)
y = df['music_genre']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.2, random_state= 42,shuffle=True,stratify=y)
```

```python
x_train
```

* Scaling

```python
x_train.columns
```

```python
from sklearn.preprocessing import StandardScaler

# Apply RobustScaler
scaler = StandardScaler()
x_train[['popularity', 'acousticness', 'danceability', 'duration_ms', 'energy',
       'instrumentalness' , 'liveness', 'loudness', 
       'speechiness', 'tempo', 'valence']] = scaler.fit_transform(x_train[['popularity', 'acousticness', 'danceability', 'duration_ms', 'energy',
       'instrumentalness' , 'liveness', 'loudness', 
       'speechiness', 'tempo', 'valence']])
x_test[['popularity', 'acousticness', 'danceability', 'duration_ms', 'energy',
       'instrumentalness' , 'liveness', 'loudness', 
       'speechiness', 'tempo', 'valence']] = scaler.transform(x_test[['popularity', 'acousticness', 'danceability', 'duration_ms', 'energy',
       'instrumentalness' , 'liveness', 'loudness', 
       'speechiness', 'tempo', 'valence']])
```

```python
df.music_genre.value_counts(normalize=True)*100
```

```python
#x_train_num
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
x_train['key'] = le.fit_transform(x_train['key'])
x_test['key'] = le.transform(x_test['key'])
```

# Machine Learning

* Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report

gauss_nb = GaussianNB()

gauss_nb.fit(x_train, y_train)

print('Train gauss_nb\n', classification_report(y_train, gauss_nb.predict(x_train)))
print('\nTest gauss_nb\n', classification_report(y_test, gauss_nb.predict(x_test)))
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_train_le = le.fit_transform(y_train)
y_test_le = le.transform(y_test)
```

```python
cat_cols_indices = x_train.columns.get_indexer(['key','mode']).tolist()
cat_cols_indices
```

```python
from mixed_naive_bayes import MixedNB

mixed_nb = MixedNB(categorical_features=cat_cols_indices)

mixed_nb.fit(x_train, y_train_le)

print('Train mixed_nb\n', classification_report(y_train_le, mixed_nb.predict(x_train)))
print('\nTest mixed_nb\n', classification_report(y_test_le, mixed_nb.predict(x_test)))
```
