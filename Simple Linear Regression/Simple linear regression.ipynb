### Ordinary Least Square (OLS)

```python
x_bar=df.area.mean()
x_bar
```

```python
y_bar=df.price.mean()
y_bar
```

```python
theta_1 = sum( (df.area - x_bar) * (df.price - y_bar) ) / sum((df.area - x_bar) ** 2)
theta_0 = y_bar - (theta_1 * x_bar)

print(theta_0)
print(theta_1)
```

```python
from sklearn.linear_model import LinearRegression

LinearRegressionModel = LinearRegression()

LinearRegressionModel.fit(df[['area']], df['price'])
```

```python
print("coefficient is : " ,LinearRegressionModel.coef_)
print("intercept is :" ,LinearRegressionModel.intercept_)
```

```python
LinearRegressionModel.predict([[1700]])
```

```python
px.scatter(data_frame= df, x= 'area', y= 'price', trendline= 'ols', trendline_color_override= 'red')
```

```python
area_predicted = LinearRegressionModel.predict(df[['area']])
area_predicted
```

```python
df['price_predicted']=area_predicted
df['price_predicted']
```

### Gradient Descent

```python
from sklearn.linear_model import SGDRegressor

sgd = SGDRegressor()

sgd.fit(df[['area']], df['price'])
```

```python
sgd.eta0
```

```python
sgd.early_stopping
```

```python
price_predicted_sgd = sgd.predict(df[['area']])
price_predicted_sgd
```

```python
df['price_predicted_sgd'] = price_predicted_sgd
```

```python
df
```

```python
fig = px.scatter(df, x='area', y='price')
fig.add_scatter(x=df['area'], y=df['price_predicted_sgd'], 
                mode='lines', line=dict(color='red'), 
                name='price_predicted_sgd').show()
```
