## Split Data Train and Test

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=46,shuffle=True,stratify=y)
```

* Handle Numerical Columns

```python
scaling_columns= x_train.columns.drop(['Male','Country'])
scaling_columns
```

```python
from sklearn.preprocessing import RobustScaler

rc=RobustScaler()
x_train[scaling_columns]=rc.fit_transform(x_train[scaling_columns])
x_test[scaling_columns]=rc.transform(x_test[scaling_columns])
```

* Handle Categorical Columns

```python
from category_encoders import BinaryEncoder

bc=BinaryEncoder()
bc_train=bc.fit_transform(x_train[['Country']])
bc_test=bc.transform(x_test[['Country']])
```

```python
x_train=pd.concat([x_train,bc_train],axis=1).drop("Country",axis=1)
x_test=pd.concat([x_test,bc_test],axis=1).drop("Country",axis=1)
```
